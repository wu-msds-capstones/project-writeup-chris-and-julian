import os import pandas as pd from sqlalchemy import create_engine

autorack = create_engine(os.environ\[“AUTORACK_URL”\])

q_scale = ““” select count(distinct e.county_fips) as counties, sum(b.population) as population, sum((e.routine_cognitive_share + e.routine_manual_share) \* b.population) / sum(b.population) as routine_task_content from county_task_exposure e join county_baseline b on e.county_fips = b.county_fips and e.year = b.year where e.year = 2023 and b.population is not null ““”

scale = pd.read_sql(q_scale, autorack) print(scale)

q_national = ““” select sum(population) as population from county_baseline where year = 2023 ““”

national = pd.read_sql(q_national, autorack) print(national)
